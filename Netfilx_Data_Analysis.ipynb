```python
import pandas as pd
import seaborn as sns
import matplotlib as plt

df = pd.read_csv("netflix_dataset.csv")
df
```

```python
print(df.info())
print(df.isnull().sum())
```

```python
print(df['type'].value_counts())
```

```python
import matplotlib.pyplot as plt

df['type'].value_counts().plot(kind='bar')

plt.title("Movies vs TV Shows")
plt.xlabel("Type")
plt.ylabel("Count")

plt.show()
```

```python
top_countries = df['country'].value_counts().head(10)

print(top_countries)
```

```python
top_countries.plot(kind='barh')

plt.title("Top 10 Countries")
plt.show()
```

```python
year_data = df['release_year'].value_counts().sort_index()

year_data.plot()

plt.title("Netflix Releases Over Years")
plt.xlabel("Year")
plt.ylabel("Count")

plt.show()
```

```python
genres = df['listed_in'].str.split(',', expand=True).stack()

print(genres.value_counts().head(10))
```

Insights:
- Movies are more common than TV Shows.
- USA produces the highest Netflix content.
- Netflix content increased rapidly after 2015.
